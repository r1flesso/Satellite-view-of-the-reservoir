Уберем SVC, так как он очень медленный, а также обучался не на всей выборке

```python
df = df[df['experiment'] != 'SVC']
df.shape
```

Унификация столбцов с метрикой F1. В разных экспериментах MLflow сохранял F1-score то в столбце f1, то в f1_score. Этот код объединяет оба столбца в один f1, беря значение из любого непустого источника

```python
df['f1'] = df['f1'].combine_first(df['f1_score'])
df['f1'].isnull().sum()
```

Удаляем ненужные или повторяющиеся столбцы, а также выводим таблицу полностью

```python
df = df.drop(columns=['experiment', 'run_id', 'recall_score', 'accuracy_score', \
                      'precision_score', 'example_count', 'score', 'log_loss', 'f1_score'])

with pd.option_context('display.max_rows', None):
    display(df)
```

Пропуски значений метрик, которые мы не логировали у некоторых моделей, заполняем нулями

```python
df = df.fillna(0)
df.isnull().sum()
```

Выбираем 4 важные метрики, нормализуем их, прописываем кастомный словарь весов важности метрик, суммируем произведение веса на нормализованную величину метрики. Выводим топ-10 самых лучших результатов новой кастомной метрики - composite_score

```python
important_metrics = ['f1', 'roc_auc', 'precision', 'recall']

df_norm = df[important_metrics].apply(lambda x: (x - x.min()) / (x.max() - x.min()))

weights_of_metrics = {
    'f1': 0.4,
    'roc_auc': 0.3,
    'precision': 0.15,
    'recall': 0.15
}

df['composite_score'] = sum(df_norm[metric] * weight for metric, weight in weights_of_metrics.items())

df = df.sort_values('composite_score', ascending=False)
df.head(10)
```

Это называется тотальная доминация Random Forest. Просто весь топ-10 - это одни леса...
Самая лучшая модель - Random Forest с max_features='log2', bootstrap=False и улучшаемой метрикой - matthews_corrcoef.
